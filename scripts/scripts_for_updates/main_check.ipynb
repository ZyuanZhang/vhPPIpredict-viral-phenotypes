## 首先看一下 更新后的 (Gao Wenze) 感染人的病毒 (high, middle and low) 相比于之前，有什么差异

对应 Figure 2 的相关内容

```python
def main_human_infecting_virus_diff():
    import pandas as pd

    dt_original = pd.read_excel("../../tables_in_paper/Table S2.xlsx", sheet_name="Table S2", header=1)
    dt_original_high = dt_original[dt_original["confidence level"]=="high"].reset_index(drop=True)
    dt_original_middle = dt_original[dt_original["confidence level"]=="middle"].reset_index(drop=True)
    dt_original_low = dt_original[dt_original["confidence level"]=="low"].reset_index(drop=True)

    dt_gwz = pd.read_excel("../../data/dataset_from_gaowenze/human_infecting_viruses_from_gwz.xlsx", sheet_name="感染人类病毒数据", header=0)
    dt_gwz_high = dt_gwz[dt_gwz["Evidence level"]=="high"].reset_index(drop=True)
    dt_gwz_middle = dt_gwz[dt_gwz["Evidence level"]=="middle"].reset_index(drop=True)
    dt_gwz_low = dt_gwz[dt_gwz["Evidence level"]=="low"].reset_index(drop=True)

    print(dt_original_high.shape, dt_gwz_high.shape)
    print(dt_original_middle.shape, dt_gwz_middle.shape)
    print(dt_original_low.shape, dt_gwz_low.shape)

    ## 为什么 high 还变少了呢？
    vtaxid_high_original = dt_original_high["virus taxid"].tolist()
    vtaxid_high_gwz = dt_gwz_high["Taxid"].tolist()
    vtaxid_middle_gwz = dt_gwz_middle["Taxid"].tolist()
    vtaxid_low_gwz = dt_gwz_low["Taxid"].tolist()

    print(len(set(vtaxid_high_original)&set(vtaxid_high_gwz)), len(set(vtaxid_high_original)&set(vtaxid_middle_gwz)), len(set(vtaxid_high_original)&set(vtaxid_low_gwz)))

    ## 看一下 FengYang 的 middle, low 在 GWZ 中是不是也是中和低
    vtaxid_middle_original = dt_original_middle["virus taxid"].tolist()
    print(len(set(vtaxid_middle_original)&set(vtaxid_high_gwz)), len(set(vtaxid_middle_original)&set(vtaxid_middle_gwz)), len(set(vtaxid_middle_original)&set(vtaxid_low_gwz)))
    print(set(vtaxid_middle_original)&set(vtaxid_high_gwz))

    vtaxid_low_original = dt_original_low["virus taxid"].tolist()
    print(len(set(vtaxid_low_original)&set(vtaxid_high_gwz)), len(set(vtaxid_low_original)&set(vtaxid_middle_gwz)), len(set(vtaxid_low_original)&set(vtaxid_low_gwz)))


    ## 这些 human-infecting viruses 与 mollentze's dataset 中的 non-human-infecting viruses 之间是否存在交集？
    vtaxid_all_human_infecting_viruses = list(set(vtaxid_high_original) | set(vtaxid_middle_original) | set(vtaxid_low_original) | \
                                              set(vtaxid_high_gwz) | set(vtaxid_middle_gwz) | set(vtaxid_low_gwz))
    dt_mollentze = pd.read_excel("../../tables_in_paper/Table S3.xlsx", sheet_name="Table S3", header=1)
    dt_mollentze_non_human_infecting = dt_mollentze[dt_mollentze["Label_text"]=="non-infecting human"].reset_index(drop=True)
    vtaxid_non_human_infecting_viruses = dt_mollentze_non_human_infecting["Taxid"].unique().tolist()

    #print(len(vtaxid_all_human_infecting_viruses), len(vtaxid_non_human_infecting_viruses), len(set(vtaxid_all_human_infecting_viruses)&set(vtaxid_non_human_infecting_viruses)))

    print(len(set(vtaxid_high_original)&set(vtaxid_non_human_infecting_viruses)), set(vtaxid_high_original)&set(vtaxid_non_human_infecting_viruses))
    print(len(set(vtaxid_middle_original)&set(vtaxid_non_human_infecting_viruses)), set(vtaxid_middle_original)&set(vtaxid_non_human_infecting_viruses))
    print(len(set(vtaxid_low_original)&set(vtaxid_non_human_infecting_viruses)),set(vtaxid_low_original)&set(vtaxid_non_human_infecting_viruses))

    print(len(set(vtaxid_high_gwz)&set(vtaxid_non_human_infecting_viruses)), set(vtaxid_high_gwz)&set(vtaxid_non_human_infecting_viruses))
    print(len(set(vtaxid_middle_gwz)&set(vtaxid_non_human_infecting_viruses)), set(vtaxid_middle_gwz)&set(vtaxid_non_human_infecting_viruses))
    print(len(set(vtaxid_low_gwz)&set(vtaxid_non_human_infecting_viruses)), set(vtaxid_low_gwz)&set(vtaxid_non_human_infecting_viruses))


if __name__ == "__main__":
    main_human_infecting_virus_diff()
```

```python
## 先暂时用全部的病毒吧 (GFY + GWZ)
def main_human_infecting_virus_merged():
    import pandas as pd

    dt_fy = pd.read_excel("../../tables_in_paper/Table S2.xlsx", sheet_name="Table S2", header=1)
    dt_fy = dt_fy.loc[:, ["virus taxid", "virus genome"]]
    dt_fy["person name"] = ["FY",]*dt_fy.shape[0]
    dt_gwz = pd.read_excel("../../data/dataset_from_gaowenze/human_infecting_viruses_from_gwz.xlsx", sheet_name="感染人类病毒数据", header=0)
    dt_gwz = dt_gwz.loc[:, ["Taxid", "Genome"]]
    dt_gwz.columns = ["virus taxid", "virus genome"]
    dt_gwz["person name"] = ["GWZ",]*dt_gwz.shape[0]

    dt_merge = pd.concat([dt_fy, dt_gwz], ignore_index=True)
    dt_merge = dt_merge.drop_duplicates(subset="virus taxid", keep="first")
    dt_merge.to_csv("../../data_updated/human_infecting_viruses_from_fy_gwz.csv", sep="\t", index=False)
    print(dt_merge.shape)
    

if __name__ == "__main__":
    main_human_infecting_virus_merged()
```
